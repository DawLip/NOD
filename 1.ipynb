```python
#46 Odchylenie standardowe. Policz odchylenie standardowe (df["kol"].std() lubnp.std). Zinterpretuj jako miarę rozproszenia danych.

print("Odchylenie standardowe: " + str(df["math score"].std()))
```

```python
#47 Wariancja. Policz wariancję (df["kol"].var() lub np.var) i pokaż, że jest kwadratem odchylenia standardowego.

print("Wariancja: " + str(df["math score"].var()))
print("Kwadrat odchylenia standardowego: " + str(df["math score"].std() ** 2))
```

```python
#48 Korelacja. Policz korelację Pearsona między dwiema kolumnami liczbowymi(df["a"].corr(df["b"]) lub np.corrcoef). Zinterpretuj znak i siłę zależności.

print("Korelacja: " + str(df["math score"].corr(df["reading score"])))

# korelacja jest silna, dodatnia
```

```python
#49 Kowariancja. Policz kowariancję tych samych dwóch kolumn (df["a"].cov(df["b"]) lub np.cov). Wyjaśnij, czym różni się od korelacji. 
print("Kowariancja: " + str(df["math score"].cov(df["reading score"])))
# Kowarancja i korelacja określają jak bardzo dwie zmienne są ze sobą powiązane oraz kierónek powiązania. 
# Korelacja przyjmuje wartości -1 do 1, a kowariancja jest zależna od jednostek miary i zmiennych
```

```python
# Na koniec zadanie podsumowujące: 
# wybierz dwie kolumny (np. dwie wielkości, które mogą być powiązane), 
# policz dla nich średnią, medianę i odchylenie standardowe, a następnie korelację i kowariancję między nimi. 
# Napisz kilka zdań wniosku: czy wartości są skupione czy rozproszone, czy między kolumnami jest zależność i jakiego kierunku.

print("reading score - średnia: " + str(df["reading score"].mean()))
print("reading score - mediana: " + str(df["reading score"].median()))
print("reading score - odchylenie standardowe: " + str(df["reading score"].std()) + "\n")

print("writing score - średnia: " + str(df["writing score"].mean()))
print("writing score - mediana: " + str(df["writing score"].median()))
print("writing score - odchylenie standardowe: " + str(df["writing score"].std()) + "\n")

print("Korelacja między reading score a writing score: " + str(df["reading score"].corr(df["writing score"])))
print("Kowariancja między reading score a writing score: " + str(df["reading score"].cov(df["writing score"])))

# Odhylenie standardowe wskazuje, że wartości są skupione wokół średniej. Korelacja 0.95 wskazuje na bardzo silną dodatnią korelację między wynikami.
```
